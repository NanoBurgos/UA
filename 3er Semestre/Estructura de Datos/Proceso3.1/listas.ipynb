# **Clasificación 1 :**

**Lineales :**  Los elementos se organizan uno detrás de otro.

- Arreglos (arrays/vectores): colección de datos del mismo tipo.
- Listas: secuencia dinámica de elementos.
- Pilas (stack): acceso LIFO (último en entrar, primero en salir).
- Colas (queue): acceso FIFO (primero en entrar, primero en salir).

**No lineales :** Los elementos se organizan jerárquica o en red.

- Árboles: estructura jerárquica con nodos y ramas.
- Grafos: conjunto de nodos conectados por aristas, útiles para representar redes.

# **Clasificación 2 :**

**Estructuras estáticas :** Su tamaño se fija en el momento de la compilación y no puede cambiar durante la ejecución del programa.

- Arreglos (arrays/vectores)
- Matrices

**Estructuras dinámicas:** Su tamaño puede variar durante la ejecución del programa, adaptándose a las necesidades

- Listas enlazadas (linked lists):  Cada nodo apunta al siguiente.
- Pilas (stack) y colas (queue) dinámicas:  Implementadas con nodos enlazados.
- Árboles y grafos:  Se construyen dinámicamente con nodos y enlaces.

**Estructuras de control vs estructuras de datos :**

Las estructuras de control deciden cómo se ejecuta el programa, mientras que las estructuras de datos deciden cómo se guarda y organiza la información.
