# **Estadística II**

## Fórmulas de referencia

**Distribución discreta:**
$$\mu=\sum xP(x),\qquad \sigma^2=\sum (x-\mu)^2P(x)=\sum x^2P(x)-\mu^2,\qquad \sigma=\sqrt{\sigma^2}.$$

**Binomial** (ensayos independientes, probabilidad de éxito constante):
$$P(X=r)=\binom nr p^r q^{n-r},\qquad q=1-p,\qquad \mu=np,\qquad \sigma^2=npq.$$

**Normal:** $$Z=\frac{X-\mu}{\sigma},\quad P(a<X<b)=\Phi(z_b)-\Phi(z_a).$$

**Intervalos de confianza (procedimiento Z utilizado en clase para muestras grandes):**
$$IC_\mu=\bar x\pm Z_{\alpha/2}\frac{s}{\sqrt n},\quad IC_p=\hat p\pm Z_{\alpha/2}\sqrt{\frac{\hat p(1-\hat p)}{n}},\quad \hat p=\frac{x}{n}.$$

**Tamaño muestral para una proporción:** $$n=\frac{Z_{\alpha/2}^2\hat p(1-\hat p)}{E^2}.$$ Redondear siempre hacia arriba.

Valores críticos: 90% → 1,645; 95% → 1,96; 98% → 2,326; 99% → 2,576; 96% → aproximadamente 2,054.

> **Precisiones:** el material de intervalos usa Z cuando $n\geq30$ aun cuando se proporciona $s$ en lugar de $\sigma$. Se mantiene ese criterio didáctico. El método es aproximado; bajo supuestos adicionales se puede emplear la distribución t para medias con desviación poblacional desconocida. Para proporciones se siguen los intervalos normales indicados por el profesor.

# Parte I — Marco teórico
## A. Selección múltiple

**1. b)** La suma de probabilidades de una distribución discreta debe ser **1**: $\sum P(x)=1$.

**2. b)** La normal estándar tiene **$\mu=0$ y $\sigma=1$**.

**3. c)** El Teorema del Límite Central justifica una aproximación normal de la media muestral para muestras suficientemente grandes (habitualmente se usa $n>30$ como regla orientativa; importan también las condiciones de muestreo y la distribución original).

**4. c)** Al aumentar el nivel de confianza aumenta $Z_{\alpha/2}$, por lo tanto el margen de error y la anchura del intervalo.

**5. b)** La distribución **normal** permite aproximar la distribución de la proporción muestral con tamaño y condiciones adecuados.

## B. Completamiento

1. Variable aleatoria **discreta**.
2. Curva normal **simétrica** respecto a la media.
3. Estimador **puntual**.
4. Valor crítico **1,96** para un nivel de confianza del 95%.
5. Una **proporción** muestral ($\hat p$).

> Las definiciones sobre distribución discreta y normal están en la Unidad II; los términos sobre estimación e intervalos corresponden al otro material y al enunciado del trabajo.

# Parte II — Ejercicios prácticos
## Tema 1. Distribución de probabilidad discreta

### Ejercicio 1 — Paquetes dañados
**Datos:** $P(0)=0,60$, $P(1)=0,25$, $P(2)=0,10$, $P(3)=0,05$.

| $x$ | 0 | 1 | 2 | 3 |
|---|---:|---:|---:|---:|
| $P(x)$ | 0,60 | 0,25 | 0,10 | 0,05 |

**a) Media:**
$$\mu=\sum xP(x)=0(0,60)+1(0,25)+2(0,10)+3(0,05)=\boxed{0,60}.$$
**Respuesta:** se esperan 0,60 paquetes dañados por día, en promedio.

**b) Al menos un paquete:**
$$P(X\geq1)=1-P(0)=1-0,60=\boxed{0,40=40\%}.$$
**Respuesta:** 40% de probabilidad.

### Ejercicio 2 — Cuatro preguntas independientes
**Datos:** $n=4$, $p=0,70$, $q=1-p=0,30$. Distribución **binomial**.

**a) Tabla completa:**
$$P(X=r)=\binom4r(0,70)^r(0,30)^{4-r}.$$

| $r$ | Sustitución | $P(X=r)$ |
|---:|---|---:|
| 0 | $\binom40(0,7)^0(0,3)^4$ | 0,0081 |
| 1 | $\binom41(0,7)^1(0,3)^3$ | 0,0756 |
| 2 | $\binom42(0,7)^2(0,3)^2$ | 0,2646 |
| 3 | $\binom43(0,7)^3(0,3)^1$ | 0,4116 |
| 4 | $\binom44(0,7)^4(0,3)^0$ | 0,2401 |
| **Total** | | **1,0000** |

**b) Varianza binomial:**
$$\sigma^2=npq=4(0,70)(0,30)=\boxed{0,84}.$$
**Respuesta:** varianza 0,84 respuestas correctas al cuadrado.

### Ejercicio 3 — Imperfecciones por lámina
**Datos:** $P(0)=0,4$, $P(1)=0,3$, $P(2)=0,2$, $P(3)=0,1$; 15 USD por imperfección.

**a) Desviación estándar:**
$$\mu=0(0,4)+1(0,3)+2(0,2)+3(0,1)=1.$$
$$\sum x^2P(x)=0^2(0,4)+1^2(0,3)+2^2(0,2)+3^2(0,1)=2.$$
$$\sigma^2=2-1^2=1,\qquad \sigma=\sqrt1=\boxed{1\text{ imperfección}}.$$

**b) Costo esperado:**
$$E(C)=15\,E(X)=15(1)=\boxed{15\text{ USD por lámina}}.$$

### Ejercicio 4 — Clientes en fila
**Datos:** $P(1)=0,15$, $P(2)=0,30$, $P(3)=0,35$, $P(4)=0,20$.

**a) Más de dos:**
$$P(X>2)=P(3)+P(4)=0,35+0,20=\boxed{0,55=55\%}.$$

**b) Valor esperado:**
$$\mu=1(0,15)+2(0,30)+3(0,35)+4(0,20)$$
$$=0,15+0,60+1,05+0,80=\boxed{2,60\text{ clientes}}.$$

### Ejercicio 5 — Ventas por hora
**Datos:** $P(0)=0,50$, $P(1)=0,35$, $P(2)=0,15$.

**a) Comprobación:** todas las probabilidades pertenecen a $[0,1]$ y
$$\sum P(x)=0,50+0,35+0,15=\boxed{1}.$$
**Respuesta:** sí es una función de probabilidad legítima (resultados excluyentes y suma igual a 1).

**b) Ventas esperadas en 8 horas:**
$$\mu=0(0,50)+1(0,35)+2(0,15)=0,65\text{ ventas/hora}.$$
$$E(\text{ventas en 8 horas})=8(0,65)=\boxed{5,20\text{ ventas}}.$$
**Interpretación:** unas 5,2 ventas en promedio por jornada, no necesariamente 5 ventas exactas.

## Tema 2. Distribución normal
Se aplica $Z=(X-\mu)/\sigma$. Las probabilidades se obtienen de la tabla normal estándar (o con la función acumulada $\Phi$).

### Ejercicio 6 — Ensamblaje de motores
**Datos:** $\mu=45$ minutos, $\sigma=6$ minutos.

**a) Más de 50 minutos:**
$$z=\frac{50-45}{6}=0,8333.$$
$$P(X>50)=1-\Phi(0,8333)=1-0,7977=\boxed{0,2023=20,23\%}.$$

**b) Entre 39 y 48 minutos:**
$$z_1=\frac{39-45}{6}=-1;\qquad z_2=\frac{48-45}{6}=0,50.$$
$$P(39<X<48)=\Phi(0,50)-\Phi(-1)=0,6915-0,1587$$
$$=\boxed{0,5328=53,28\%}.$$

### Ejercicio 7 — Salarios de técnicos
**Datos:** $\mu=22$ USD/hora, $\sigma=3,50$ USD/hora.

**a) Menos de 18 USD/hora:**
$$z=\frac{18-22}{3,50}=-1,1429.$$
$$P(X<18)=\Phi(-1,1429)=\boxed{0,1265\approx12,65\%}.$$

**b) Salario que separa al 10% mejor pagado:** percentil 90, $\Phi(z)=0,90$, por tanto $z\approx1,2816$.
$$X=\mu+z\sigma=22+1,2816(3,50)=\boxed{26,49\text{ USD/hora}}.$$
**Interpretación:** aproximadamente el 10% de los técnicos gana más de ese importe.

### Ejercicio 8 — Prueba de aptitud
**Datos:** $\mu=500$ puntos y $\sigma=100$ puntos.

**a) Entre 450 y 620:**
$$z_1=\frac{450-500}{100}=-0,50;\qquad z_2=\frac{620-500}{100}=1,20.$$
$$P(450<X<620)=\Phi(1,20)-\Phi(-0,50)$$
$$=0,8849-0,3085=\boxed{0,5764=57,64\%}.$$

**b) Top 15%:** equivale al percentil 85; $z_{0,85}\approx1,0364$.
$$X=500+1,0364(100)=\boxed{603,64\text{ puntos}}.$$
**Respuesta:** si las calificaciones son enteras, se requieren al menos **604 puntos**.

### Ejercicio 9 — Consumo de agua
**Datos:** $\mu=18$ m³, $\sigma=4,2$ m³.

**a) Más de 25 m³:**
$$z=\frac{25-18}{4,2}=1,6667.$$
$$P(X>25)=1-\Phi(1,6667)\approx1-0,9522=\boxed{0,0478=4,78\%}.$$

**b) Menos de 15 m³ en 500 hogares:**
$$z=\frac{15-18}{4,2}=-0,7143.$$
$$P(X<15)=\Phi(-0,7143)\approx0,2375.$$
$$500(0,2375)=118,75\approx\boxed{119\text{ hogares}}.$$

### Ejercicio 10 — Vida útil de baterías
**Datos:** $\mu=4,5$ años, $\sigma=0,8$ años.

**a) Duración entre 3,5 y 5,2 años:**
$$z_1=\frac{3,5-4,5}{0,8}=-1,25;\quad z_2=\frac{5,2-4,5}{0,8}=0,875.$$
$$P(3,5<X<5,2)=\Phi(0,875)-\Phi(-1,25)$$
$$\approx0,8092-0,1056=\boxed{0,7036=70,36\%}.$$

**b) Garantía para que solo el 3% falle antes:** buscar el percentil 3, $z_{0,03}\approx-1,8808$.
$$G=\mu+z\sigma=4,5-1,8808(0,8)=2,9954\text{ años}.$$
$$\boxed{G\approx3,00\text{ años}}.$$
**Interpretación:** el plazo corresponde aproximadamente al tercer percentil de vida útil, bajo la distribución normal supuesta.

## Tema 3. Intervalos de confianza para la media (muestras grandes)
Procedimiento del material *Intervalos de Confianza (Media y Proporción)*: identificar $n$, $\bar x$, $s$ y confianza; calcular $E=Z_{\alpha/2}s/\sqrt n$ y luego $IC=[\bar x-E,\bar x+E]$.

### Ejercicio 11 — Gasto promedio
**Datos:** $n=64$, $\bar x=85$ USD, $s=12$ USD, confianza 95%, $Z=1,96$.

**a) Intervalo:**
$$E=1,96\frac{12}{\sqrt{64}}=1,96(1,5)=2,94.$$
$$IC=85\pm2,94=\boxed{[82,06;\ 87,94]\text{ USD}}.$$
**Interpretación:** el procedimiento produce un intervalo con 95% de confianza para el verdadero gasto promedio poblacional.

**b) Si $n$ aumenta a 150** manteniendo $\bar x$ y $s$:
$$E_{150}=1,96\frac{12}{\sqrt{150}}\approx1,9204.$$
$$IC_{150}\approx85\pm1,9204=\boxed{[83,08;\ 86,92]}.$$
**Comparación:** la amplitud pasa de $2(2,94)=5,88$ a $2(1,9204)\approx3,84$. Una muestra mayor reduce el error y estrecha el intervalo.

### Ejercicio 12 — Embotellamientos
**Datos:** $n=49$, $\bar x=52$ min, $s=10,5$ min, 99%, $Z=2,576$.

**a) Intervalo:**
$$E=2,576\frac{10,5}{\sqrt{49}}=2,576(1,5)=3,864.$$
$$IC=52\pm3,864=\boxed{[48,136;\ 55,864]\text{ minutos}}.$$
A dos decimales: **[48,14; 55,86] minutos**.

**b) Margen de error máximo:** $\boxed{E=3,864\text{ minutos}\approx3,86\text{ min}}$.

### Ejercicio 13 — Bombillas LED
**Datos:** $n=100$, $\bar x=12.500$ horas, $s=800$ horas, 90%, $Z=1,645$.

**a) Intervalo:**
$$E=1,645\frac{800}{\sqrt{100}}=1,645(80)=131,60\text{ h}.$$
$$IC=12.500\pm131,60=\boxed{[12.368,40;\ 12.631,60]\text{ h}}.$$

**b) ¿Se respalda la afirmación de 12.700 horas?** **No**: 12.700 es mayor que el límite superior 12.631,60. La afirmación no resulta compatible con este intervalo de confianza del 90%.

### Ejercicio 14 — Autoestudio
**Datos:** $n=81$, $\bar x=14,2$ horas, $s=3,6$ horas, 95%, $Z=1,96$.

**a) Intervalo:**
$$E=1,96\frac{3,6}{\sqrt{81}}=1,96(0,4)=0,784\text{ h}.$$
$$IC=14,2\pm0,784=\boxed{[13,416;\ 14,984]\text{ h}}.$$
A dos decimales: **[13,42; 14,98] horas**.

**b) Reducir el margen de error a la mitad:** como $E\propto1/\sqrt n$ con lo demás constante, es necesario cuadruplicar el tamaño muestral:
$$n_2=4n_1=4(81)=\boxed{324\text{ estudiantes}}.$$

### Ejercicio 15 — Peso de porciones
**Datos:** $n=36$, $\bar x=340$ g, $s=15$ g, 98%, $Z\approx2,326$.

**a) Intervalo:**
$$E=2,326\frac{15}{\sqrt{36}}=2,326(2,5)=5,815\text{ g}.$$
$$IC=340\pm5,815=\boxed{[334,185;\ 345,815]\text{ g}}.$$
A dos decimales: **[334,19; 345,82] gramos** (redondeo usual).

**b) Interpretación de $n=36$ y TLC:** la muestra supera el umbral orientativo de 30 indicado en el material. El Teorema del Límite Central permite aproximar la distribución de la media muestral mediante una normal para muestras suficientemente grandes, suponiendo muestreo adecuado e independencia. No significa que todos los pesos individuales sean normales.

## Tema 4. Intervalos de confianza para la proporción
Procedimiento del material: $\hat p=x/n$, $q=1-\hat p$, $EE=\sqrt{\hat p q/n}$, $E=Z_{\alpha/2}EE$, $IC=[\hat p-E,\hat p+E]$.

### Ejercicio 16 — Preferencia por productos orgánicos
**Datos:** $n=400$, $x=180$, 95%, $Z=1,96$.

**a) Estimación puntual e intervalo:**
$$\hat p=\frac{180}{400}=0,45=45\%;\qquad q=0,55.$$
$$EE=\sqrt{\frac{0,45(0,55)}{400}}\approx0,024875.$$
$$E=1,96(0,024875)\approx0,048754.$$
$$IC=0,45\pm0,048754=\boxed{[0,4012;\ 0,4988]}$$
**Respuesta:** **[40,12%; 49,88%]** aproximadamente.

**b) Muestra para error máximo $E=0,02$:**
$$n=\frac{(1,96)^2(0,45)(0,55)}{(0,02)^2}=2376,99.$$
Redondeando hacia arriba, $\boxed{n=2377\text{ consumidores}}$.

### Ejercicio 17 — Componentes defectuosos
**Datos:** $n=150$, $x=12$, confianza 99%, $Z=2,576$.

**a) Intervalo:**
$$\hat p=12/150=0,08;\qquad q=0,92.$$
$$EE=\sqrt{\frac{0,08(0,92)}{150}}\approx0,022151.$$
$$E=2,576(0,022151)\approx0,05706.$$
$$IC=0,08\pm0,05706\approx\boxed{[0,0229;\ 0,1371]}.$$
**Respuesta:** **[2,29%; 13,71%]**.

**b) Aproximación normal:**
$$n\hat p=150(0,08)=12\geq5;\qquad n(1-\hat p)=150(0,92)=138\geq5.$$
Se cumplen las dos condiciones usadas en el enunciado para la aproximación normal.

### Ejercicio 18 — Intención de voto
**Datos:** $n=600$, $x=312$, 90%, $Z=1,645$.

**a) Intervalo:**
$$\hat p=312/600=0,52;\qquad q=0,48.$$
$$EE=\sqrt{\frac{0,52(0,48)}{600}}\approx0,020396.$$
$$E=1,645(0,020396)\approx0,03355.$$
$$IC=0,52\pm0,03355\approx\boxed{[0,4865;\ 0,5535]}.$$
**Respuesta:** **[48,65%; 55,35%]**.

**b) ¿Se puede asegurar que superará el 50%?** **No**, porque el intervalo incluye el 50% y valores inferiores. Además, se trata de una estimación de intención de voto, no de resultados definitivos.

### Ejercicio 19 — Satisfacción del servicio
**Datos:** $n=250$, $x=215$, 95%, $Z=1,96$.

**a) Intervalo:**
$$\hat p=215/250=0,86;\qquad q=0,14.$$
$$EE=\sqrt{\frac{0,86(0,14)}{250}}\approx0,021945.$$
$$E=1,96(0,021945)\approx0,04301.$$
$$IC=0,86\pm0,04301\approx\boxed{[0,8170;\ 0,9030]}.$$
**Respuesta:** **[81,70%; 90,30%]**.

**b) Error estándar:** $\boxed{EE\approx0,02195}$, equivalente a aproximadamente **2,195 puntos porcentuales**.

### Ejercicio 20 — Cobertura de vacunación
**Datos:** $n=500$, $x=385$, confianza 96%.

**a) Intervalo:**
$$\hat p=385/500=0,77;\qquad q=0,23.$$
Para 96% de confianza, $\alpha=0,04$ y el valor crítico deja 0,98 a la izquierda: $Z\approx2,054$.
$$EE=\sqrt{\frac{0,77(0,23)}{500}}\approx0,018820.$$
$$E=2,054(0,018820)\approx0,03865.$$
$$IC=0,77\pm0,03865\approx\boxed{[0,7313;\ 0,8087]}.$$
**Respuesta:** aproximadamente **[73,13%; 80,87%]**.

**b) Meta de 80%:** el valor 80% está dentro del intervalo y la proporción estimada es solo del 77%. **No es posible afirmar que la cobertura poblacional alcanzó la meta del 80%** con este intervalo.
